```python
#Converting array of strings in List

def convertgenres(stri):
    import ast
    l = []
    for i in ast.literal_eval(stri):
        l.append(i['name'])
    return l


#Calling the function  
movies['genres'] = movies['genres'].apply(convertgenres)


# Note: Once converted into list. if we again run these function it will show error because we had saved the converted list into the same column and if we call again the function these time we are passing the converted list to the function again. which will show error
```

```python

#printing movies['genres'] after converting
print(movies['genres'])
```

```python
#same we also want to fetch only name from keywords columns 
movies['keywords'] = movies['keywords'].apply(convertgenres)
```

```python

movies.head()
```

```python
# defining function to remove only first 3 main character names from the movie.
#Cast column contain id,name, and more info of all characters from the movie. each movies in one list. i just want to fetch first 3 characters name's from that list

def castandcrewconvert(stri):
    import ast
    l = []
    counter = 0
    for i in ast.literal_eval(stri):
        if counter != 3:
            l.append(i['name'])
            counter+=1
        else:
            break
    return l
            
            
```

```python
movies['cast'] = movies['cast'].apply(castandcrewconvert)
```

```python
# Written a function to fetch only director from the column key, remaining data in not required for model
def convertcast(stri):
    import ast
    l = []
    for i in ast.literal_eval(stri):
        if i['job'] == "Director":
            l.append(i['name'])
            break
    return l

# Call the function and Saved the Result in crew column
movies['crew'] = movies['crew'].apply(convertcast)        
```

```python
#Checked out the result
movies.sample(2)
```

```python
# overview column contains string which we need to convert to list. for that we had done these. currently i don't know why we had done it
movies["overview"] = movies["overview"].apply(lambda x  : x.split())
```

```python
movies.sample()
```

```python
# Removed space from the texts because for eg. sam in cast column and sam in crew column can be represented as same by model though they are different if removing space can make on sambahadur and another samkaran model can understand it cleary 
movies['genres'] = movies['genres'].apply(lambda x : [i.replace(" ","")for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x : [i.replace(" ","")for i in x])
movies['cast'] = movies['cast'].apply(lambda x : [i.replace(" ","")for i in x])
movies['crew'] = movies['crew'].apply(lambda x : [i.replace(" ","")for i in x])
```

```python
movies.head()
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
new_df = movies[["movie_id","title","tags"]]
```

```python
new_df.head()
```

```python
#Converting List to String
new_df['tags'] = new_df['tags'].apply(lambda x : " ".join(x))
```

```python
new_df['tags'][0]
```

```python
#Converted Strings each letter into LowerCase
new_df['tags'] = new_df['tags'].apply(lambda x : x.lower())
```

```python
new_df.head()
```

```python
# We want to bring words into its base form for that we had import stemmer from nltk library
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python

# We had created a function which accepts a string, break it into seperate words, then bring words to its base form, append the words into a list (y) and then again create a string from the list and return it
def stem(text):
    y = []

    for i in text.split():   #it split the string into seperate word
        y.append(ps.stem(i))  #ps.stem --> it bring the word into its base form    ----- y.append append those base word into list
    return " ".join(y)        # convert list into string by joining each word with spaces and return the string   



# if we want to perform stemming only one word directly write ps.stem("insert word") no need to write the whole function

```

```python

new_df['tags'] = new_df['tags'].apply(stem)    #Call the stem function and passed tags column in it
new_df.head()   # Checked for the result
```

```python

# Converted text to Vectors because computer doesnt understand text. it only understand numbers. there are more methods to handle categorial data also

from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(max_features=5000, stop_words="english")  #max 5000 features are taken means 1st para --> row1 == 5000 words --> cols., 2nd para --> row2 5000 words -->cols.
newvector = vectorizer.fit_transform(new_df['tags']).toarray()
```

```python

#printing vector shape and which words are converted into vectors. why it print 5000. because 5000 words for each para represented as column for each row
newvector.shape
print(vectorizer.get_feature_names_out().tolist())
```

```python
# Imported Cosine_similarity beacuse we can to calculate cosine distance from one movie to another. 
# The less the distance the more similar to other movie is to it. thhe more the distance the less similar the other movie is to it.
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Called function cosine_similarity to calculate distance
distanc = cosine_similarity(newvector)

# Try to print it by just removing cs = from the code
# You will realize that -- for 1st movie-- 1st index is 1 because 1st movies has same(equal) distance from itself.
#                       -- for 2nd movie-- 2nd index is 1 bacause 2nd movie has same(equal) distance from itself.
#                       -- Similary the diagonal has the 1 value and other had the calculated distance. 
```
